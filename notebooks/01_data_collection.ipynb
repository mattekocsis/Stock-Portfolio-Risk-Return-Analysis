```python
import yfinance as yf
import pandas as pd

tickers = ["AAPL", "MSFT", "NVDA", "JNJ", "UNH", "XOM", "CVX", "PG", "KO", "SPY"]
data = yf.download(tickers, start="2020-01-01", end="2025-12-31", auto_adjust=True)["Close"]
data.to_csv("prices_raw.csv")
```

These tickers were chosen based on sector diversification, a large market cap and liquidity as well as depth of data history with decades of clean daily data and dividend policy variation.

Python code snippet 'auto_adjust=True' automatically adjusts for dividend and stock splits which is a requirement for accurate return calculations.
